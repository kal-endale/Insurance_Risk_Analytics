```python
import pandas as pd

# Load the insurance data into a pandas DataFrame
df = pd.read_csv(r'D:\KAIM-9\Week-3\Insurance_Risk_Analytics\data\Data\insurance_data.csv')


# Display the first 5 rows to inspect the data
display(df.head())
```

# Task-1

```python
print('DataFrame Shape:', df.shape)
print('\n--- Data Info ---')
df.info()
print('\n--- Missing Values ---')
display(df.isnull().sum())
print('\n--- Descriptive Statistics ---')
display(df.describe(include='all'))
```

```python
# Calculate Loss Ratio
df['LossRatio'] = df['TotalClaims'] / df['TotalPremium']

# Calculate Margin
df['Margin'] = df['TotalPremium'] - df['TotalClaims']

# Convert 'TransactionDate' to datetime objects
df['TransactionDate'] = pd.to_datetime(df['TransactionDate'])

# Display the DataFrame head with the new columns and check data types again
display(df.head())
df.info()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select numerical columns for histograms
numerical_cols = [
    'Age', 'AnnualIncome', 'RiskScore', 'AnnualPremium', 'Deductible',
    'NCD', 'PastClaims', 'ClaimAmount', 'TotalPremium', 'TotalClaims',
    'LossRatio', 'Margin'
]

# Create histograms for numerical features
plt.figure(figsize=(18, 15))
for i, col in enumerate(numerical_cols):
    plt.subplot(4, 3, i + 1)  # Adjust subplot grid as needed
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
# Select categorical columns for bar charts (excluding CustomerID and ZipCode for now)
categorical_cols = [
    'Gender', 'Province', 'VehicleType', 'Claimed', 'CoverType', 'AutoMake'
]

plt.figure(figsize=(18, 12))
for i, col in enumerate(categorical_cols):
    plt.subplot(2, 3, i + 1)  # Adjust subplot grid as needed
    sns.countplot(y=df[col], order=df[col].value_counts().index, palette='viridis')
    plt.title(f'Count of {col}')
    plt.xlabel('Count')
    plt.ylabel(col)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='TotalPremium', y='TotalClaims', data=df, alpha=0.6)
plt.title('TotalPremium vs TotalClaims')
plt.xlabel('Total Premium')
plt.ylabel('Total Claims')
plt.grid(True)
plt.show()

# Calculate correlation between TotalPremium and TotalClaims
correlation = df['TotalPremium'].corr(df['TotalClaims'])
print(f"Correlation between TotalPremium and TotalClaims: {correlation:.2f}")
```

```python
# Group by ZipCode and calculate average premium, claims, loss ratio, and margin
zipcode_summary = df.groupby('ZipCode').agg(
    AvgPremium=('TotalPremium', 'mean'),
    AvgClaims=('TotalClaims', 'mean'),
    TotalPolicies=('ZipCode', 'size'),
    TotalClaimed=('Claimed', lambda x: (x == True).sum())
).reset_index()

# Add Average Loss Ratio (careful with division by zero)
zipcode_summary['AvgLossRatio'] = zipcode_summary['AvgClaims'] / zipcode_summary['AvgPremium']
zipcode_summary.loc[zipcode_summary['AvgPremium'] == 0, 'AvgLossRatio'] = 0 # Handle cases where AvgPremium is 0

# Display summary
display(zipcode_summary.sort_values(by='AvgLossRatio', ascending=False).head())

# Visualize AvgLossRatio by ZipCode
plt.figure(figsize=(12, 6))
sns.barplot(x='ZipCode', y='AvgLossRatio', data=zipcode_summary.sort_values('AvgLossRatio', ascending=False))
plt.title('Average Loss Ratio by ZipCode (Top ZipCodes by Loss Ratio)')
plt.xlabel('Zip Code')
plt.ylabel('Average Loss Ratio')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.scatterplot(x='AvgPremium', y='AvgClaims', hue='ZipCode', size='TotalPolicies', sizes=(20, 400), data=zipcode_summary)
plt.title('Average Premium vs Average Claims by ZipCode')
plt.xlabel('Average Premium')
plt.ylabel('Average Claims')
plt.grid(True)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 7))
sns.countplot(x='Province', hue='CoverType', data=df, palette='pastel')
plt.title('Cover Type Distribution by Province')
plt.xlabel('Province')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.legend(title='Cover Type')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16, 8))
sns.barplot(x='Province', y='AnnualPremium', hue='AutoMake', data=df, palette='viridis', ci=None)
plt.title('Average Annual Premium by Province and Auto Make')
plt.xlabel('Province')
plt.ylabel('Average Annual Premium')
plt.xticks(rotation=45)
plt.legend(title='Auto Make', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Select key numerical columns for outlier detection
key_numerical_cols = [
    'Age', 'AnnualIncome', 'RiskScore', 'AnnualPremium', 'Deductible',
    'NCD', 'PastClaims', 'ClaimAmount', 'LossRatio', 'Margin', 'CustomValueEstimate'
]

plt.figure(figsize=(18, 15))
for i, col in enumerate(key_numerical_cols):
    plt.subplot(4, 3, i + 1) # Adjust subplot grid as needed
    sns.boxplot(y=df[col])
    plt.title(f'Box Plot of {col}')
    plt.ylabel(col)
plt.tight_layout()
plt.show()
```

```python
# Calculate overall Loss Ratio for the portfolio
overall_total_claims = df['TotalClaims'].sum()
overall_total_premium = df['TotalPremium'].sum()
overall_loss_ratio = overall_total_claims / overall_total_premium
print(f"Overall Loss Ratio for the portfolio: {overall_loss_ratio:.2f}")

# Loss Ratio by Province
loss_ratio_by_province = df.groupby('Province').apply(lambda x: x['TotalClaims'].sum() / x['TotalPremium'].sum()).reset_index(name='LossRatio')
print("\nLoss Ratio by Province:")
display(loss_ratio_by_province.sort_values('LossRatio', ascending=False))

# Loss Ratio by VehicleType
loss_ratio_by_vehicletype = df.groupby('VehicleType').apply(lambda x: x['TotalClaims'].sum() / x['TotalPremium'].sum()).reset_index(name='LossRatio')
print("\nLoss Ratio by VehicleType:")
display(loss_ratio_by_vehicletype.sort_values('LossRatio', ascending=False))

# Loss Ratio by Gender
loss_ratio_by_gender = df.groupby('Gender').apply(lambda x: x['TotalClaims'].sum() / x['TotalPremium'].sum()).reset_index(name='LossRatio')
print("\nLoss Ratio by Gender:")
display(loss_ratio_by_gender.sort_values('LossRatio', ascending=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(18, 6))

# Plot Loss Ratio by Province
plt.subplot(1, 3, 1)
sns.barplot(x='Province', y='LossRatio', data=loss_ratio_by_province.sort_values('LossRatio', ascending=False), palette='viridis')
plt.title('Loss Ratio by Province')
plt.xlabel('Province')
plt.ylabel('Loss Ratio')
plt.xticks(rotation=45, ha='right')

# Plot Loss Ratio by VehicleType
plt.subplot(1, 3, 2)
sns.barplot(x='VehicleType', y='LossRatio', data=loss_ratio_by_vehicletype.sort_values('LossRatio', ascending=False), palette='magma')
plt.title('Loss Ratio by Vehicle Type')
plt.xlabel('Vehicle Type')
plt.ylabel('Loss Ratio')
plt.xticks(rotation=45, ha='right')

# Plot Loss Ratio by Gender
plt.subplot(1, 3, 3)
sns.barplot(x='Gender', y='LossRatio', data=loss_ratio_by_gender.sort_values('LossRatio', ascending=False), palette='cividis')
plt.title('Loss Ratio by Gender')
plt.xlabel('Gender')
plt.ylabel('Loss Ratio')

plt.tight_layout()
plt.show()
```

```python
# Ensure TransactionDate is datetime type and set as index for time series analysis
df_time = df.set_index('TransactionDate').sort_index()

# Resample data by month to analyze trends
df_monthly = df_time.resample('ME').agg(
    ClaimFrequency=('Claimed', 'size'), # Count of policies each month (using size of grouped entries)
    TotalClaimsAmount=('TotalClaims', 'sum'), # Total claim amount each month
    TotalPolicies=('CustomerID', 'nunique') # Assuming CustomerID is unique per policy
)

# Calculate Claim Frequency (policies with claims / total policies) and Claim Severity (total claims / policies with claims)
df_monthly['ClaimFrequencyRatio'] = df_monthly['ClaimFrequency'] / df_monthly['TotalPolicies']
df_monthly['AverageClaimAmount'] = df_monthly['TotalClaimsAmount'] / df_monthly['ClaimFrequency']

print("Monthly Trends in Claim Frequency and Severity:")
display(df_monthly.head())

# Plotting temporal trends
plt.figure(figsize=(15, 10))

plt.subplot(3, 1, 1)
plt.plot(df_monthly.index, df_monthly['ClaimFrequency'], marker='o', linestyle='-')
plt.title('Monthly Claim Count')
plt.ylabel('Number of Claims')
plt.grid(True)

plt.subplot(3, 1, 2)
plt.plot(df_monthly.index, df_monthly['TotalClaimsAmount'], marker='o', linestyle='-', color='orange')
plt.title('Monthly Total Claim Amount (Severity)')
plt.ylabel('Total Claim Amount')
plt.grid(True)

plt.subplot(3, 1, 3)
plt.plot(df_monthly.index, df_monthly['ClaimFrequencyRatio'], marker='o', linestyle='-', color='green')
plt.title('Monthly Claim Frequency Ratio (Claims / Total Policies)')
plt.ylabel('Claim Frequency Ratio')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Group by AutoMake and VehicleModel to calculate average ClaimAmount
claim_by_make_model = df.groupby(['AutoMake', 'VehicleModel'])['ClaimAmount'].mean().reset_index()

# Sort to find highest average claim amounts
highest_claims_make_model = claim_by_make_model.sort_values(by='ClaimAmount', ascending=False).head(10)
print("Top 10 Vehicle Makes/Models with Highest Average Claim Amounts:")
display(highest_claims_make_model)

# Sort to find lowest average claim amounts
lowest_claims_make_model = claim_by_make_model.sort_values(by='ClaimAmount', ascending=True).head(10)
print("\nTop 10 Vehicle Makes/Models with Lowest Average Claim Amounts:")
display(lowest_claims_make_model)

# Visualize the top 5 highest and lowest claim amounts by make/model
plt.figure(figsize=(15, 6))

plt.subplot(1, 2, 1)
sns.barplot(x='ClaimAmount', y='VehicleModel', hue='AutoMake', data=highest_claims_make_model.head(5), palette='Reds_d', dodge=False)
plt.title('Top 5 Vehicle Makes/Models by Highest Average Claim Amount')
plt.xlabel('Average Claim Amount')
plt.ylabel('Vehicle Model')
plt.legend(title='Auto Make', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.subplot(1, 2, 2)
sns.barplot(x='ClaimAmount', y='VehicleModel', hue='AutoMake', data=lowest_claims_make_model.head(5), palette='Greens_d', dodge=False)
plt.title('Top 5 Vehicle Makes/Models by Lowest Average Claim Amount')
plt.xlabel('Average Claim Amount')
plt.ylabel('Vehicle Model')
plt.legend(title='Auto Make', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```
